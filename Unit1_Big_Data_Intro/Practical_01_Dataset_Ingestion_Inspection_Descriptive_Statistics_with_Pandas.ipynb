# Practical 01: Dataset Ingestion, Inspection & Descriptive Statistics with Pandas

**UNIT 1: Introduction to Big Data and Large Dataset Handling**

**Problem Statement:** Write a Python program using Pandas to load a structured dataset (CSV/JSON), display its first 5 rows, last 5 rows, dimensions (shape), column names, and data types with info(). Generate comprehensive summary statistics of all numeric and categorical attributes using describe(include='all').

```python
import io
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None, "display.width", 200)

rg = np.random.default_rng(99)
n = 220
original = pd.DataFrame({
    "Order ID": np.arange(1001, 1001 + n),
    "Customer Name": [f"Buyer_{i:03d}" for i in rg.integers(1, 55, n)],
    "City": rg.choice(["Kolkata", "Patna", "Ranchi", "Delhi", "Lucknow", "Surat"], n),
    "Product Category": rg.choice(["Electronics", "Clothing", "Groceries", "Furniture", "Toys"], n),
    "Quantity": rg.integers(1, 10, n),
    "Unit Price": np.round(rg.uniform(120, 5500, n), 2),
    "Discount %": rg.choice([0, 5, 10, 15, 20], n),
    "Order Date": pd.date_range("2025-04-10", periods=n, freq="D").strftime("%Y-%m-%d"),
})

csv_stream = io.StringIO()
original.to_csv(csv_stream, index=False)
csv_stream.seek(0)
json_stream = io.StringIO(original.to_json(orient="records"))

frame = pd.read_csv(csv_stream)
print("Dataset loaded successfully!")

print(frame.head())

print(frame.tail())

print("Shape (rows, columns):", frame.shape)
print("Column names:", frame.columns.tolist())

frame.info()

print(frame.describe(include='all'))

df_js = pd.read_json(json_stream)
print("JSON load – shape:", df_js.shape)
print(df_js.head())
```
