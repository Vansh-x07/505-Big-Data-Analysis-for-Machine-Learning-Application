# Practical 16: Automated Text Summarization using Pretrained LLM Pipeline

**UNIT 4: Introduction to Large Language Models and Big Data Applications**

**Problem Statement:** Write a Python program to perform automated text summarization using a Hugging Face pretrained transformer model. Initialize the pipeline('summarization'), pass a multi-sentence news or technical article, specify max_length and min_length parameters, and generate a concise abstractive summary.

```python
from transformers import pipeline

sum_model = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

document = """
Big data technologies have transformed the way organisations store, process and analyse
information. Traditional tools struggle when datasets grow beyond the memory of a single
machine, so engineers adopt distributed frameworks such as Apache Spark that split work across
many cores or servers. At the same time, columnar file formats like Parquet reduce storage
costs and speed up analytical queries by reading only the columns that are needed.
Once data is cleaned and prepared, machine learning models can be trained to forecast demand,
detect fraud or recommend products. More recently, large language models trained on huge text
corpora have made it possible to summarise documents, answer questions and analyse sentiment
with very little task-specific training. Companies now combine scalable data pipelines with
pretrained language models to extract insights from the vast amount of unstructured text they
collect every day.
"""
print(document.strip())

generated = sum_model(document.strip(), max_length=60, min_length=25, do_sample=False)
abstract = generated[0]["summary_text"]
print(abstract)

print(f"Original words: {len(document.split())} | Summary words: {len(abstract.split())}")
```
