# Practical 14: Vocabulary Frequency Profiling & Stopword Filtering

**UNIT 4: Introduction to Large Language Models and Big Data Applications**

**Problem Statement:** Write a Python program to profile vocabulary in a text corpus: clean sentences, filter out common English stopwords (e.g., 'the', 'is', 'in', 'and'), calculate word occurrence frequencies using collections.Counter, and display the top 5 most frequent keywords.

```python
import string
from collections import Counter

corpus = [
    "Hospitals generate large volumes of data from scans and patient records.",
    "The data is cleaned and anonymised before it is used to train a model.",
    "Distributed systems process big data on many machines at the same time.",
    "Language models can summarise long reports into short and useful notes.",
    "Doctors and data scientists work together to improve patient care.",
]

IGNORE = {"the", "a", "an", "is", "are", "of", "to", "for", "on", "at", "by", "in", "from", "and", "can", "into", "before", "with", "same", "time"}

def clean(sentence: str) -> str:
    sentence = sentence.lower()
    for p in string.punctuation:
        sentence = sentence.replace(p, "")
    return " ".join(sentence.split())

corpus_words = []
for s in corpus:
    corpus_words.extend(clean(s).split())

meaningful = [w for w in corpus_words if w not in IGNORE]

print(f"Total tokens (before filtering): {len(corpus_words)}")
print(f"Total tokens (after  filtering): {len(meaningful)}")
print(f"Vocabulary size (unique words) : {len(set(meaningful))}")

tally = Counter(meaningful)
print(tally)

for rank, (word, count) in enumerate(tally.most_common(7), 1):
    print(f"{rank}. {word:<10} {count}  {'#' * count}")
```
