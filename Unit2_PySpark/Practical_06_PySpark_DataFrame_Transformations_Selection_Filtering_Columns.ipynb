# Practical 06: PySpark DataFrame Transformations: Selection, Filtering & Columns

**UNIT 2: Big Data Processing using PySpark**

**Problem Statement:** Write a PySpark program to execute DataFrame transformations: select specific columns using .select(), filter rows using multi-condition boolean logic (e.g., Salary > 60000 and City == 'Surat') using .filter(), create a computed column using .withColumn(), and rename a column using .withColumnRenamed().

```python
from pyspark.sql import SparkSession, Row
from pyspark.sql.functions import col

spark = (SparkSession.builder.appName("DS505-Lab06").master("local[*]")
         .config("spark.ui.showConsoleProgress", "false").getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

data = [
    {"Name": "Rutvik", "Department": "IT", "City": "Nagpur", "Salary": 73000},
    {"Name": "Hetal", "Department": "HR", "City": "Indore", "Salary": 55000},
    {"Name": "Dhruv", "Department": "Finance", "City": "Nagpur", "Salary": 68000},
    {"Name": "Mansi", "Department": "IT", "City": "Bhopal", "Salary": 89000},
    {"Name": "Chirag", "Department": "Sales", "City": "Indore", "Salary": 50000},
    {"Name": "Kavya", "Department": "IT", "City": "Nagpur", "Salary": 94000},
]
employees = spark.createDataFrame([Row(**d) for d in data])
employees.show()

employees.select("Name", "City", "Salary").show()

filtered = employees.filter((col("Salary") > 60000) & (col("City") == "Nagpur"))
filtered.show()

with_bonus = employees.withColumn("Annual_Bonus", col("Salary") * 0.15)
with_bonus.show()

renamed = with_bonus.withColumnRenamed("Name", "Employee_Name")
renamed.show()
renamed.printSchema()
spark.stop()
```
