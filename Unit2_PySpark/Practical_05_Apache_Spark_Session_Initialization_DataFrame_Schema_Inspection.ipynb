# Practical 05: Apache Spark Session Initialization & DataFrame Schema Inspection

**UNIT 2: Big Data Processing using PySpark**

**Problem Statement:** Write a Python program to install and initialize an Apache Spark session in PySpark using SparkSession.builder. Create a structured PySpark DataFrame from Python dictionaries, display its schema tree hierarchy using printSchema(), and print its rows using .show().

```python
from pyspark.sql import SparkSession, Row

spark = (SparkSession.builder
         .appName("DS505-Lab05")
         .master("local[*]")
         .config("spark.ui.showConsoleProgress", "false")
         .getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

print("Spark version:", spark.version)

data = [
    {"Name": "Rutvik", "Department": "IT", "City": "Nagpur", "Salary": 73000, "Experience": 5},
    {"Name": "Hetal", "Department": "HR", "City": "Indore", "Salary": 55000, "Experience": 3},
    {"Name": "Dhruv", "Department": "Finance", "City": "Nagpur", "Salary": 68000, "Experience": 7},
    {"Name": "Mansi", "Department": "IT", "City": "Bhopal", "Salary": 89000, "Experience": 8},
    {"Name": "Chirag", "Department": "Sales", "City": "Indore", "Salary": 50000, "Experience": 2},
]
employees = spark.createDataFrame([Row(**d) for d in data])
employees.printSchema()
employees.show()
spark.stop()
```
